```python
import pandas as pd
import numpy as np

# 1. Read the dataset
df = pd.read_csv('lab_data_chaotic.csv')

# 2. Fill the NaN parts with the median of numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# 3. Clear ages higher than 100 and lower than 0
if 'Age' in df.columns:
    df = df[(df['Age'] >= 0) & (df['Age'] <= 100)]

# 4. Drop the Eye Color column
if 'Eye_Color' in df.columns:
    df = df.drop(columns=['Eye_Color'])

# 5. Anonymize the Patient_ID column
if 'Patient_ID' in df.columns:
    df['Patient_ID'] = '***'

# 6. Convert the Test_Date column to datetime format (coerce invalid to NaT)
if 'Test_Date' in df.columns:
    df['Test_Date'] = pd.to_datetime(df['Test_Date'], errors='coerce')

# 7. Filter dataset to keep only rows where status is 'Completed'
if 'Status' in df.columns:
    df = df[df['Status'] == 'Completed']

# 8. Print the first 5 lines
print("Cleaned and Anonymized DataFrame:")
display(df.head())
```

It loads 'lab_data_chaotic.csv', fills missing values with their respective column medians, filters out invalid ages (keeping only those between 0 and 100), drops the 'Eye_Color' column, and displays the first 5 rows.

Data Loading: Loaded the dataset successfully.
Handling Missing Values: Replaced all missing (NaN) values in numeric columns (like Age and Hemoglobin_Level) with their respective medians.
Age Cleaning: Kept only valid age records strictly between 0 and 100 years old.
Column Dropping: Completely removed the unnecessary Eye_Color column.
Anonymization: Replaced all identifiers in the Patient_ID column with '***' for privacy.
Datetime Conversion: Converted the Test_Date column into a proper Datetime format.
Filtering: Retained only the rows where the status is 'Completed'.
The resulting output displays the first 5 clean, filtered, and anonymized rows of your bio-data.
